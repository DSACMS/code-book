```python
import pandas as pd
import numpy as np

# Using triple quotes prevents the internal single/double quotes from breaking the f-string
print(f"""{"Variable":<20} | {"NaN/None":<10} | {"Spaces (' ')":<12} | {"Empty ('')":<10} | {"Dots ('.')":<10}""")
print("-" * 70)

# Loop through every column in your dataset
for col in df.columns:
    nans = df[col].isna().sum()
    
    # Safely check for string patterns (handles numeric columns without breaking)
    spaces = (df[col] == ' ').sum() if df[col].dtype == 'object' else 0
    empties = (df[col] == '').sum() if df[col].dtype == 'object' else 0
    dots = (df[col] == '.').sum() if df[col].dtype == 'object' else 0
    
    # ONLY print if the variable has at least one missing value type present
    if nans > 0 or spaces > 0 or empties > 0 or dots > 0:
        print(f"{col:<20} | {nans:<10} | {spaces:<12} | {empties:<10} | {dots:<10}")
```

```python
updated_formats = meta.original_variable_types.copy()
```
